## 🧠 Tuesday — How LLMs Work: Probability & Prediction
### Concept Recap
- LLMs are trained via **maximum likelihood estimation**: predict the next token given context.
- They model \( P(w_t \mid w_1, ..., w_{t-1}) \).  
- This explains prompt sensitivity and **hallucinations** — outputs are drawn from a probability distribution, not factual reasoning.

### 🧩 Classroom Demo — Next Token Sampling
We’ll simulate next-token probabilities and show how sampling works using Python.

````python
import numpy as np
import matplotlib.pyplot as plt

# Hypothetical token probabilities (e.g., next word predictions)
tokens = ['cat', 'dog', 'robot', 'car', 'pizza']
probs = np.array([0.4, 0.3, 0.15, 0.1, 0.05])

# Visualize probability distribution
plt.bar(tokens, probs)
plt.title('Next-Token Probability Distribution')
plt.ylabel('P(token)')
plt.show()
````

### 🧮 Sampling Simulation
Now, sample multiple times and observe diversity of outputs.

````python
np.random.seed(42)
samples = np.random.choice(tokens, size=20, p=probs)
print(samples)
unique, counts = np.unique(samples, return_counts=True)
for u, c in zip(unique, counts):
    print(f'{u}: {c}')
````

### Discussion Questions
- Why might the model sometimes produce unlikely tokens (like *pizza*)?
- What would happen if the distribution were more uniform?

# 🔥 Thursday — Temperature, Entropy & Prompt Patterns
### Concept Overview
- **Temperature** controls randomness in sampling.
  - High temperature → more exploration, creativity.
  - Low temperature → more conservative, repetitive outputs.
- **Entropy** quantifies output diversity.
- In practice: tuning temperature ≈ tuning creativity vs. reliability.

````python
def softmax(logits, temperature=1.0):
    logits = np.array(logits) / temperature
    e = np.exp(logits - np.max(logits))
    return e / e.sum()

logits = [3.0, 2.0, 0.5]
temps = [0.5, 1.0, 2.0]

for T in temps:
    probs_T = softmax(logits, temperature=T)
    print(f'Temperature {T}: {probs_T}')
````

### 🧩 Prompt Experiment
Let's demonstrate real temperature effects via the API.

````python
import openai, os
from getpass import getpass

os.environ['OPENAI_API_KEY'] = getpass('🔑 Enter your API key: ')
client = openai.OpenAI(api_key=os.environ['OPENAI_API_KEY'])

prompt = 'Write a one-sentence poem about robots learning to dream.'

for temp in [0.2, 0.7, 1.2]:
    print(f'\nTemperature = {temp}\n')
    response = client.chat.completions.create(
        model='gpt-4o-mini',
        messages=[{'role': 'user', 'content': prompt}],
        temperature=temp
    )
    print(response.choices[0].message.content)
````
