## 🧭 Day 1 (Jan 13): What is AI Engineering?

**Guiding question:**  
> What does it mean to engineer an AI system, not just build a model?

---

### 1. Welcome & Framing (0–10 min)
- **Goal:** Ground students in *engineering mindset* vs *research mindset*.
- AI Engineering = *building reliable, modular, and measurable systems that use AI models as components*.
- Research asks *“Can we make it work?”*; Engineering asks *“Will it work reliably and safely in the world?”*

Reflection prompt: *What other engineered systems (bridges, aircraft, software) face reliability–performance tradeoffs?*

---

### 2. Inspiration: AI Success Stories (10–20 min)
**Demo:** GPT-4 explaining a complex topic clearly.

```python
from openai import OpenAI
client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a teaching assistant helping students learn about AI."},
        {"role": "user", "content": "Explain how a neural network recognizes handwriting, in simple terms."}
    ]
)

print(response.choices[0].message.content)
```

*Instructor discussion:* note how LLMs can generalize and communicate complex ideas.

---

### 3. Failure Cases (20–35 min)
Even strong models **hallucinate** — produce confident falsehoods.

**Example prompt:**  
> “Explain how whales evolved from fish.”

Students run this through the API and note factual errors or overconfidence.

**Discussion:**  
- Why might the model generate confident nonsense?  
- What parts of the system could we “engineer” to reduce the risk? (e.g., validation, retrieval, monitoring)

---

### 4. The LLM System Mental Model (35–50 min)

Let’s unpack how text flows through an AI system.

This represents the *entire AI system pipeline* — not just the model.

| Layer | Purpose | Example |
|-------|----------|----------|
| **User Interaction** | How inputs are collected, formatted, filtered | Chat UI, API call |
| **Prompt & Control** | Instructions, examples, constraints | “You are a helpful tutor.” |

### 5. Live Demo: Your First OpenAI Call (50–60 min)
Hands-on walkthrough: running a short prompt and observing model behavior.

```python
# @title Run your first OpenAI API call
from openai import OpenAI

client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a helpful teaching assistant."},
        {"role": "user", "content": "Explain what AI engineering means in one sentence."}
    ]
)

print(response.choices[0].message.content)
```

**Checkpoint discussion:**  
- What does the system *not* tell us about how it reached its answer?

## 🧩 Day 2 (Jan 15): How Do LLMs Work (Conceptually)?

**Guiding question:**  
> What’s happening between your text input and the model’s output?

### Recap (0–10 min)
- Yesterday: LLMs succeed spectacularly — and sometimes fail spectacularly.
- Today: peek inside the *model* itself (tokens → probabilities → words).

### Conceptual Flow (10–20 min)
```
Word → Token → Vector → Predicted next token → Text Output
```

Analogy: The model is a “next-word prediction engine,” not a database of facts.

---

### 🧪 Python Demo: Translation Task (20–40 min)
```python
# @title Translation example
from openai import OpenAI

client = OpenAI()

prompt = "Translate this sentence into French: 'The AI engineer designs safe systems.'"
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a language tutor."},
        {"role": "user", "content": prompt}
    ]
)

print(response.choices[0].message.content)
```

- Observe that LLMs are *probabilistic translators*.
- Reflection: How might errors appear in multilingual or domain-specific contexts?
