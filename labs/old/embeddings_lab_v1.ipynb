## 🔬 Scientific Process

**Question:**

**Hypothesis:**  
I expect that semantic search will find relevant results even when no words overlap, while keyword search will only match literal terms.

- *Optional numeric metric:* cosine similarity between query and top document.

**Expected Conclusion Shape:**  
As embeddings encode semantics, we should observe higher relevance in semantic search even without shared words.


## Part 1 – Inspecting Embeddings

Let's begin by generating embeddings for a few short texts and measuring how similar they are.


```
# @title 🧠 Generate Embeddings
from openai import OpenAI
import numpy as np

client = OpenAI()
texts = ["Tulane University", "New Orleans", "banana"]
embeddings = [client.embeddings.create(input=t, model="text-embedding-3-small").data[0].embedding for t in texts]

print("Embedding length:", len(embeddings[0]))
cosine = lambda a, b: np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
print("Cosine similarity (Tulane vs New Orleans):", cosine(embeddings[0], embeddings[1]))
```
