### (15–30 min) Concept: LLMs as Probabilistic Models

An LLM predicts the next token (word or subword) based on context.

\[
P(w_t \mid w_1, \dots, w_{t-1}) = \frac{e^{s(w_t)}}{\sum_j e^{s(w_j)}}
\]

where \( s(w_j) \) is a score for each token (computed internally by the model).

Let's simulate this idea with a simple example.

```python
import numpy as np
from scipy.special import softmax

# Mock token logits (scores) for a vocabulary
tokens = ["cat", "dog", "banana", "quantum"]
scores = np.array([2.0, 1.0, 0.5, -1.0])

# Convert scores to probabilities
probs = softmax(scores)

for t, p in zip(tokens, probs):
    print(f"{t:>8s}: {p:.3f}")
```

---

### (30–45 min) Visualizing Probabilities

Let's visualize the token probabilities to make this intuition clearer.

```python
import matplotlib.pyplot as plt

plt.bar(tokens, probs)
plt.title("Next-Token Probability Distribution")
plt.ylabel("P(token | context)")
plt.show()
```

💡 *Observation:* The model is much more confident in “cat” than “banana” or “quantum.”

### (45–65 min) Visualizing the System — Unifying Diagram v0

```python
import matplotlib.pyplot as plt
import networkx as nx

G = nx.DiGraph()
G.add_edges_from([("User", "Prompt"), ("Prompt", "Model"), ("Model", "Output")])

pos = {"User": (0, 0), "Prompt": (1, 0), "Model": (2, 0), "Output": (3, 0)}

plt.figure(figsize=(8, 2))
nx.draw(G, pos, with_labels=True, node_size=4000, node_color="#b3cde3", font_size=12, arrows=True)
plt.title("Unifying System Diagram v0")
plt.show()
```

### Simulating Latency and Rate Limits

```python
import time, random

def simulate_api_call():
    latency = random.uniform(0.2, 1.2)
    print(f"Sending request... (latency ≈ {latency:.2f}s)")
    time.sleep(latency)
    print("Response received!")

simulate_api_call()
```

💡 *Observation:* Each call has real-world cost and delay.

---

### (45–60 min) Safe Usage & Cost Awareness

The cost of an API call depends on tokens used (input + output).

\[
\text{Cost} = \frac{N_{tokens}}{1000} \times r_{\text{per 1K}}
\]

```python
def estimate_cost(num_calls=10, avg_tokens=300, rate_per_1k=0.002):
    total_tokens = num_calls * avg_tokens
    cost = total_tokens / 1000 * rate_per_1k
    print(f"Estimated total cost: ${cost:.4f} for {num_calls} calls")
    return cost

estimate_cost()
```

💬 *Discussion:* How do rate limits, latency, and cost together shape engineering tradeoffs?
