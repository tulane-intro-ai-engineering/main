### 🧩 What LLMs Are / Aren’t
- A **Large Language Model (LLM)** predicts the *next token* given previous context.
- It is a *probability model over text sequences*, not a reasoning brain.
- **Strengths:** generalization, fluency, pattern completion.
- **Weaknesses:** hallucination, brittleness, lack of grounding.

# 💻 Thursday — Hello, LLM (Colab & API Demo)
**Topics:** APIs, Colab, keys, safe usage, cost awareness

We’ll now demonstrate the `User → Prompt → Model → Output` system in code.

````python
import openai, os
from getpass import getpass

os.environ['OPENAI_API_KEY'] = getpass('🔑 Enter your API key (demo use only): ')
client = openai.OpenAI(api_key=os.environ['OPENAI_API_KEY'])
print('✅ API client initialized — ready for demo.')
````

````python
prompt = 'Explain large language models in one sentence.'
response = client.chat.completions.create(
    model='gpt-4o-mini',
    messages=[{'role': 'user', 'content': prompt}],
)
print('🧠 Prompt:', prompt, '\n')
print('💬 Model Output:\n', response.choices[0].message.content)
````

````python
prompts = [
    "Explain large language models like I'm five.",
    "Explain large language models for a data scientist.",
    "Explain large language models using a cooking analogy."
]

for p in prompts:
    print(f'\n🧠 Prompt: {p}\n')
    r = client.chat.completions.create(
        model='gpt-4o-mini',
        messages=[{'role': 'user', 'content': p}],
    )
    print(r.choices[0].message.content)
````
