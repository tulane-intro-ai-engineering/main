## 🧭 Day 1 (Jan 13): What is AI Engineering?

**Guiding question:**  
> What does it mean to engineer an AI system, not just build a model?

### 1. Welcome & Framing (0–10 min)
AI Engineering is about **building reliable, safe, and explainable systems** that use AI models as components.  
Think of it as **applied systems engineering for AI** — where reliability, monitoring, and iteration matter as much as accuracy.

Reflection: *Where have you seen AI fail in everyday life (e.g., autocomplete, recommendation systems)?*


````
### 2. Inspiration: AI Success Stories (10–20 min)

Let's see an example of an LLM explaining a technical topic.

```python
from openai import OpenAI
client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a teaching assistant helping students learn about AI."},
        {"role": "user", "content": "Explain how a neural network recognizes handwriting, in simple terms."}
    ]
)
print(response.choices[0].message.content)
```
````

````
### 3. Failure Cases (20–35 min)
Even powerful models hallucinate — producing confident falsehoods.

Try running this prompt and evaluate accuracy.

```python
from openai import OpenAI
client = OpenAI()

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": "Explain how whales evolved from fish."}
    ]
)
print(response.choices[0].message.content)
```

**Discuss:** Why did the model sound confident but wrong?  
How could engineers mitigate this (retrieval, citations, post-checks)?
````

### 4. The LLM System Mental Model (35–50 min)

AI systems are **pipelines** — not just models.

![LLM System Diagram](sandbox:/mnt/data/A_flowchart_diagram_in_digital_format_illustrates_.png)

| Layer | Purpose | Example |
|-------|----------|----------|

## 🧩 Day 2 (Jan 15): How Do LLMs Work and How Do APIs Work?

**Guiding question:**  
> What happens when you send a prompt to the OpenAI API?

### 1. Recap & Discussion (0–10 min)
Yesterday, we looked at *systems thinking* for AI.  
Today, we’ll zoom into **how models talk to us via APIs**.


### 2. How APIs Work (10–25 min)

When you use `client.chat.completions.create()`, you’re sending an **API request** to a remote model, and receiving a JSON response.

![API Request Response](sandbox:/mnt/data/A_diagram_created_by_the_Assistant_depicts_a_basic.png)

Key points:
- The **client** (your Colab notebook) sends a text request.
- The **server** (OpenAI endpoint) returns a structured response.
- This is stateless — every request includes all necessary context.


### 3. What a Prompt Actually Does (25–45 min)

Prompts guide the model’s behavior. They define **context, tone, and goals**.

```python
from openai import OpenAI
client = OpenAI()

prompt = "List three ways AI can improve education."

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are an optimistic futurist."},
        {"role": "user", "content": prompt}
    ]
)
print(response.choices[0].message.content)
```

Try modifying the *system message* to change style or depth.


````
### 4. Tokenization and Prediction (45–60 min)

LLMs predict one token at a time based on probability.  

```python
from openai import OpenAI
client = OpenAI()

prompt = "Translate this sentence into French: 'The AI engineer designs safe systems.'"
response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "system", "content": "You are a language tutor."},
        {"role": "user", "content": prompt}
    ]
)
print(response.choices[0].message.content)
```

Discussion: How can probabilistic behavior lead to inconsistency or “hallucination”?
````
